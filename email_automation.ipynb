```python
import smtplib
import os
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart
from email.mime.base import MIMEBase
from email import encoders
from datetime import datetime

SENDER_EMAIL = "youremail@gmail.com"
APP_PASSWORD  = "your-app-password-here"

def send_email(to, subject, body, attachment_path=None):
    """
    Sends an email with optional file attachment.
    """
    msg = MIMEMultipart()
    msg['From'] = SENDER_EMAIL
    msg['To']   = to
    msg['Subject'] = subject
    msg.attach(MIMEText(body, 'plain'))

    # Attach a file if provided
    if attachment_path and os.path.exists(attachment_path):
        with open(attachment_path, 'rb') as f:
            part = MIMEBase('application', 'octet-stream')
            part.set_payload(f.read())
            encoders.encode_base64(part)
            filename = os.path.basename(attachment_path)
            part.add_header('Content-Disposition', f'attachment; filename={filename}')
            msg.attach(part)
            print(f"Attached: {filename}")

    with smtplib.SMTP_SSL('smtp.gmail.com', 465) as server:
        server.login(SENDER_EMAIL, APP_PASSWORD)
        server.send_message(msg)
        print(f"Sent to {to} at {datetime.now().strftime('%H:%M:%S')}")


def send_bulk_emails(recipients, subject, body_template):
    """
    Sends personalised emails to a list of recipients.
    recipients = [{'name': 'Alice', 'email': 'alice@test.com'}, ...]
    body_template = 'Hi {name}, ...'  — use {name} as a placeholder
    """
    for person in recipients:
        personalised_body = body_template.format(name=person['name'])
        send_email(
            to=person['email'],
            subject=subject,
            body=personalised_body
        )

# ── TEST (sends to yourself) ───────────────────────────────
send_email(
    to=SENDER_EMAIL,   # send to yourself as a test
    subject="Python automation test",
    body=f"This was sent automatically at {datetime.now().strftime('%d %B %Y, %H:%M')}.\n\nYour email automation script works!"
)
```
