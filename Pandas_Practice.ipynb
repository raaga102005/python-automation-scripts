```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
```

```python
# Exercise 1: Create and explore a DataFrame

data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "Age": [22, 25, 23, 28, 21],
    "City": ["Bangalore", "Mumbai", "Delhi", "Chennai", "Hyderabad"],
    "Salary": [30000, 45000, 38000, 52000, 27000]
}

df = pd.DataFrame(data)

print("=== DataFrame ===")
print(df)

print("\n=== Shape (rows, columns) ===")
print(df.shape)

print("\n=== Column names ===")
print(df.columns.tolist())

print("\n=== Data types ===")
print(df.dtypes)

print("\n=== First 3 rows ===")
print(df.head(3))

print("\n=== Basic stats ===")
print(df.describe())
```

```python
# Exercise 2: Select, filter, sort

# Select one column
print("=== Ages only ===")
print(df["Age"])

# Select multiple columns
print("\n=== Name and Salary ===")
print(df[["Name", "Salary"]])

# Filter: salary above 35000
print("\n=== Salary > 35000 ===")
high_earners = df[df["Salary"] > 35000]
print(high_earners)

# Filter: age under 24
print("\n=== Age < 24 ===")
young = df[df["Age"] < 24]
print(young)

# Sort by salary (highest first)
print("\n=== Sorted by Salary (descending) ===")
sorted_df = df.sort_values("Salary", ascending=False)
print(sorted_df)
```

```python
# Exercise 3: New columns + missing data

# Add a new column: Tax (10% of salary)
df["Tax"] = df["Salary"] * 0.10
df["Net_Salary"] = df["Salary"] - df["Tax"]

print("=== DataFrame with Tax and Net Salary ===")
print(df)

# Now simulate missing data
df_missing = df.copy()
df_missing.loc[1, "Salary"] = None
df_missing.loc[3, "City"] = None

print("\n=== DataFrame with missing values ===")
print(df_missing)

print("\n=== Check for nulls ===")
print(df_missing.isnull().sum())

# Fill missing salary with average
avg_salary = df_missing["Salary"].mean()
df_missing["Salary"].fillna(avg_salary, inplace=True)

# Drop rows where City is null
df_missing.dropna(subset=["City"], inplace=True)

print("\n=== After cleaning missing data ===")
print(df_missing)
```

```python
# Exercise 4: Groupby

data2 = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Fiona", "George"],
    "Department": ["Engineering", "Marketing", "Engineering", "HR", "Marketing", "HR", "Engineering"],
    "Salary": [50000, 35000, 48000, 30000, 37000, 32000, 52000],
    "Experience_Years": [3, 2, 4, 1, 2, 3, 5]
}

df2 = pd.DataFrame(data2)

print("=== Original Data ===")
print(df2)

# Average salary per department
print("\n=== Avg Salary by Department ===")
print(df2.groupby("Department")["Salary"].mean())

# Count employees per department
print("\n=== Employee Count by Department ===")
print(df2.groupby("Department")["Name"].count())

# Multiple aggregations at once
print("\n=== Multiple Stats by Department ===")
print(df2.groupby("Department").agg(
    Avg_Salary=("Salary", "mean"),
    Max_Salary=("Salary", "max"),
    Total_Experience=("Experience_Years", "sum")
))
```

```python
# Exercise 4: Groupby

data2 = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Fiona", "George"],
    "Department": ["Engineering", "Marketing", "Engineering", "HR", "Marketing", "HR", "Engineering"],
    "Salary": [50000, 35000, 48000, 30000, 37000, 32000, 52000],
    "Experience_Years": [3, 2, 4, 1, 2, 3, 5]
}

df2 = pd.DataFrame(data2)

print("=== Original Data ===")
print(df2)

# Average salary per department
print("\n=== Avg Salary by Department ===")
print(df2.groupby("Department")["Salary"].mean())

# Count employees per department
print("\n=== Employee Count by Department ===")
print(df2.groupby("Department")["Name"].count())

# Multiple aggregations at once
print("\n=== Multiple Stats by Department ===")
print(df2.groupby("Department").agg(
    Avg_Salary=("Salary", "mean"),
    Max_Salary=("Salary", "max"),
    Total_Experience=("Experience_Years", "sum")
))
```

```python
# Exercise 6: Mini project — Student Grade Analyzer

students = {
    "Student": ["Raaga", "Arun", "Priya", "Kiran", "Sneha", "Rohan", "Meena"],
    "Math": [88, 72, 95, 60, 78, 85, 91],
    "Science": [76, 80, 88, 55, 92, 70, 85],
    "English": [90, 65, 82, 70, 88, 75, 93],
    "History": [70, 78, 80, 65, 74, 82, 77]
}

df3 = pd.DataFrame(students)

# Calculate average score per student
df3["Average"] = df3[["Math", "Science", "English", "History"]].mean(axis=1)

# Assign grade based on average
def assign_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 60:
        return "C"
    else:
        return "D"

df3["Grade"] = df3["Average"].apply(assign_grade)

print("=== Student Report Card ===")
print(df3)

# Who scored highest overall?
top_student = df3.loc[df3["Average"].idxmax(), "Student"]
print(f"\n🏆 Top student: {top_student}")

# Subject-wise class average
print("\n=== Class Average per Subject ===")
print(df3[["Math", "Science", "English", "History"]].mean())

# How many students got each grade?
print("\n=== Grade Distribution ===")
print(df3["Grade"].value_counts())

# Save report to CSV
df3.to_csv("student_report.csv", index=False)
print("\n✅ Report saved to student_report.csv")
```
