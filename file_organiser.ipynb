```python
import os
import shutil

def organise_folder(source_folder):
    """
    Organises files in a folder into subfolders by file type.
    Example: all .pdf files go into source_folder/PDFs/
    """

    # Define categories: extension → folder name
    categories = {
        '.pdf':  'PDFs',
        '.docx': 'Documents',
        '.doc':  'Documents',
        '.xlsx': 'Spreadsheets',
        '.csv':  'Spreadsheets',
        '.png':  'Images',
        '.jpg':  'Images',
        '.jpeg': 'Images',
        '.mp4':  'Videos',
        '.py':   'Python_Scripts',
        '.txt':  'Text_Files',
    }

    # Count what we move
    moved = 0
    skipped = 0

    # Loop through every file in the source folder
    for filename in os.listdir(source_folder):
        filepath = os.path.join(source_folder, filename)

        # Skip if it's a folder (not a file)
        if os.path.isdir(filepath):
            continue

        # Get the file extension (e.g. '.pdf')
        _, ext = os.path.splitext(filename)
        ext = ext.lower()   # normalise .PDF and .pdf to the same thing

        # Find which category it belongs to
        folder_name = categories.get(ext, 'Others')   # default: 'Others'

        # Create the destination folder if it doesn't exist
        dest_folder = os.path.join(source_folder, folder_name)
        os.makedirs(dest_folder, exist_ok=True)

        # Move the file
        dest_path = os.path.join(dest_folder, filename)
        shutil.move(filepath, dest_path)
        print(f"Moved: {filename} → {folder_name}/")
        moved += 1

    print(f"\nDone. Moved {moved} files.")

# ── TEST IN COLAB ────────────────────────
# Create a test folder with dummy files first
test_folder = '/content/test_files'
os.makedirs(test_folder, exist_ok=True)

# Create dummy files of different types
for fname in ['report.pdf', 'data.csv', 'budget.xlsx',
              'photo.jpg', 'notes.txt', 'script.py']:
    with open(os.path.join(test_folder, fname), 'w') as f:
        f.write('dummy content')

print("Test files created:", os.listdir(test_folder))

# Now organise them
organise_folder(test_folder)

# Check the result
for folder in os.listdir(test_folder):
    folder_path = os.path.join(test_folder, folder)
    if os.path.isdir(folder_path):
        print(f"\n{folder}/")
        for f in os.listdir(folder_path):
            print(f"  {f}")
```
